```python
import torch
import torch.nn as nn
from torchvision import datasets
```

```python
class MyModel(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1 = nn.Linear(784,128)
    self.layer2 = nn.Linear(128,10)

  def forward(self, x):
    x = torch.relu(self.layer1(x))
    x = self.layer2(x)
    return x
```

```python
model = MyModel()
model.parameters()

for parameter in model.parameters():
  print(parameter.shape)
```

torch.Size([128, 784])   → layer1 weights


torch.Size([128])        → layer1 bias



torch.Size([10, 128])    → layer2 weights


torch.Size([10])         → layer2 biasFor layer1 = nn.Linear(784, 128):

128 neurons
each neuron has 784 incoming weights
therefore weight tensor = (128, 784)
one bias per neuron = (128,)

```python
for name, parameter in model.named_parameters():
  print(name, parameter.shape)
  print(name, parameter.grad)
```

```python
x =torch.randn(4, 784)
y = torch.tensor([0, 1, 2, 3])
print(x.shape)
print(y.shape)
```

```python
loss_fn = nn.CrossEntropyLoss()
logits = model(x)
loss = loss_fn(logits, y)
print(logits.shape)
print(loss)
loss.backward()
```
